```python
class SimpleMambaLayer(nn.Module):
    def __init__(self, d_model, d_state):
        super().__init__()
        self.D = d_model
        self.N = d_state

        # Input projection → split into SSM input and gate
        self.in_proj = nn.Linear(d_model, 2 * d_model)

        # Selective SSM parameter projections
        self.delta_proj = nn.Linear(d_model, d_model)
        self.B_proj     = nn.Linear(d_model, d_model * d_state)
        self.C_proj     = nn.Linear(d_model, d_model * d_state)

        # Diagonal A (one per channel, per state)
        self.A = nn.Parameter(-torch.rand(d_model, d_state))

        # Output projection
        self.out_proj = nn.Linear(d_model, d_model)

    def forward(self, x):
        B, L, D = x.shape

        # ---- 1. Input projection ----
        u, gate = self.in_proj(x).chunk(2, dim=-1)
        gate = torch.sigmoid(gate)                 # (B, L, D)

        # ---- 2. Selective parameters ----
        delta = nn.functional.softplus(self.delta_proj(u))     # (B, L, D)
        B_t   = self.B_proj(u).view(B, L, D, self.N)
        C_t   = self.C_proj(u).view(B, L, D, self.N)

        # ---- 3. SSM scan (sequential for clarity) ----
        s = torch.zeros(B, D, self.N, device=x.device)               # initial state
        outputs = []

        for t in range(L):
            expA = torch.exp(delta[:, t].unsqueeze(-1) * self.A)
            s = expA * s + B_t[:, t]                # state update
            y = (C_t[:, t] * s).sum(dim=-1)         # (B, D)
            outputs.append(y)

        y = torch.stack(outputs, dim=1)             # (B, L, D)

        # ---- 4. Gating + output ----
        y = y * gate
        y = self.out_proj(y)

        return y
```

```python
class HMV(torch.nn.Module):
  def __init__(self, state_dim, embed_dim,input_dim,seq_len):
    super().__init__()
    self.state_dim = state_dim
    self.embed_dim = embed_dim
    self.input_dim = input_dim
    self.mlp=torch.nn.Sequential(
        torch.nn.Linear(self.input_dim,self.embed_dim),
        torch.nn.GELU(),
        torch.nn.Linear(self.embed_dim,self.embed_dim),
        torch.nn.GELU(),
        torch.nn.Linear(self.embed_dim,self.embed_dim))
    self.cls=torch.nn.Parameter(torch.randn(1,self.embed_dim))
    self.masktoken=torch.nn.Parameter(torch.randn(1,self.input_dim))
    self.ans0=torch.nn.Parameter(torch.randn(1,self.embed_dim))

    self.mamba1=SimpleMambaLayer(self.embed_dim,self.state_dim)
    self.mamba2=SimpleMambaLayer(self.embed_dim,self.state_dim)
    self.mamba3=SimpleMambaLayer(self.embed_dim,self.state_dim)
    self.decoder=torch.nn.Sequential( torch.nn.Linear(self.embed_dim,self.embed_dim),
        torch.nn.GELU(),
        torch.nn.Linear(self.embed_dim,self.input_dim),
                                    )
    self.transnorm=torch.nn.LayerNorm(self.embed_dim)
    self.pos_embedding = torch.nn.Parameter(torch.randn(1,seq_len , embed_dim))

  def forward(self,x,maskidx,breaktil,total):
      #(B,P,L)
    x_safe=x.clone()
    if maskidx is not None:
      for idx,i in enumerate(maskidx):
          #print(x_safe[idx,i].shape)
          x_safe[idx,i]=self.masktoken.expand(i.shape[0],-1)
    
    vecimg=self.mlp(x_safe) + self.pos_embedding
      
    cls=self.cls.expand(x_safe.shape[0],1,-1)
    ans=self.ans0.expand(x_safe.shape[0],1,-1)
    #print(vecimg.shape)  
    tokens=torch.cat([vecimg,cls,ans],dim=1)
    
    for i in range(total):
      #print(tokens.shape)
      h=tokens
      h=self.mamba1(h)
      h=self.mamba2(h)

      delta=self.mamba3(h)[:,-1,:]
      delta=self.transnorm(delta).reshape(x_safe.shape[0],1,-1)
      ans=ans+delta

      if i<breaktil:
        ans=ans.detach()
     

      tokens=torch.cat([vecimg,cls,ans],dim=1)

    ans=self.decoder(ans)

    return ans







```
